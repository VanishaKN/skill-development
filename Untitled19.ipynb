```python
marks = int(input("Enter marks: "))
if marks >= 90:
        grade = "A+"
elif marks >= 80:
        grade = "A"
elif marks >= 70:
       grade = "B"
elif marks >= 60:
       grade = "C"
elif marks >= 50:
        grade = "D"
else:
     grade = "F"

if grade in ("A+", "A"):
     status = "Distinction"
elif grade == "F":
     status = "Fail"
else:
    status= "Pass"

print(f"Grade : {grade}")
print(f"Status : {status}")
```

```python
#Program 2

balance = float(input("Enter current balance : "))
amount = float(input("Enter withdrawal amount : "))
account_type = int(input("Enter account type(1 = savings, 2=current) : "))

if amount > 50000:
    print("EXCEEDS MAXIMUM LIMIT")
elif amount % 100 != 0:
    print("NOT A MULTIPLE OF 100")
elif account_type == 1: #savings
    if balance - amount < 500:
        print("INSUFFICIENT BALANCE")
    else
       balance -= amount
       print(f"Success. New Balance : {balance : .2f}")
else: #current
    fee = 50 if amount > 25000 else 0
    balance = balance-amount-fee
    if fee>0:
         print(f"Success. Fee charged : {fee : .2f}, New Balance : {balance : .2f}")
    else:
         print(f"Success. New Balance : {balance : .2f}")
```
